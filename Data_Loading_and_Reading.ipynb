```python
# 🔎 Data Acquisition & Filtering
```

```python
## 1. Analyzing Resource Type Categories
```

```python
## 4. Assessing Access Mode and Device Categories
```

```python
df["access_mode"].unique()
print(df["access_mode"].value_counts())

df["device"].unique()
print(df["device"].value_counts())
```

```python
## 5. Selecting Required Columns
```

```python
selected_data = df[
    [
        "access_id",
        "user_id",
        "user_type",
        "department",
        "resource_id",
        "resource_title",
        "resource_type",
        "publisher",
        "access_datetime",
        "session_minutes",
        "device",
        "downloaded",
        "rating",
        "access_mode",
    ]
]
display(selected_data.head())
```

```python
## 6. Filtering Downloaded Resources
```

```python
downloaded_cases = df[df["downloaded"] == 1]
display(downloaded_cases)
print("Number of downloads: ", downloaded_cases.shape)
```

```python
## 7. Filtering Remote Access Cases
```

```python
remote_cases = df[df["access_mode"] == "Remote"]
display(remote_cases)
print("Shape of Remote Cases:", remote_cases.shape)
```

```python
## 8. Filtering Mobile Device Cases
```

```python
mobile_cases = df[df["device"] == "Mobile"]
display(mobile_cases)
print("Shape of Mobile Cases:", mobile_cases.shape)
```

```python
## 9. Filtering Faculty Usage
```

```python
faculty_cases = df[df["user_type"] == "Faculty"]
display(faculty_cases)
print("Number of faculty_cases: ", faculty_cases.shape)
```

```python
## 10. Filtering Highly Rated Resources
```

```python
high_rated = df[df["rating"] >= 4]
display(high_rated)
print("Number of high rated accesses: ", high_rated.shape)
```

```python
## 11. Filtering CSE Department Cases
```

```python
cse_cases = df[df["department"] == "CSE"]
display(cse_cases)
print("Number of cse_cases : ", cse_cases.shape)
```

```python
# Filtering Cases with Rating Information
```

```python
rated_cases = df[df["rating"].notna()]

display(rated_cases)
print("Shape of Rated Cases:", rated_cases.shape)
```
